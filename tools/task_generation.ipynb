```python
# create info.json and task config
import os
import json
scene_folder = os.path.join(os.path.expanduser("~"), "data", "isaac_scenes_v1")

template = \
"""      {scene_name}:
        label: {label}
        ceiling_height: 1.0
        episode_number: 20
        navmesh_exclude:
        - /World/ground/terrain/__default_setting/HDR_Sphere/SM_Sphere
        navmesh_preset: default
        path: grscenes_commercial/scenes/{scene_name}/start_result_navigation.usd
        rule_pattern: gr"""

data_folder = os.path.join(scene_folder, "grscenes_commercial", "scenes")

# count = 0
# for scene_name in os.listdir(data_folder):
#     if os.path.isdir(os.path.join(data_folder, scene_name)):
#         scene_label = "none"
#         new_scene_name = f'scene{count}'
#         os.rename(os.path.join(data_folder, scene_name), os.path.join(data_folder, new_scene_name))
#         count += 1

data = []
for scene_name in os.listdir(data_folder):
    if os.path.isdir(os.path.join(data_folder, scene_name)):
        if os.path.exists(os.path.join(data_folder, scene_name, "info.json")):
            info = json.load(open(os.path.join(data_folder, scene_name, "info.json")))
            scene_label = info["label"]
            if scene_label[-1].isdigit():
                scene_label = scene_label[:-1]
            with open(os.path.join(data_folder, scene_name, "info.json"), "w") as f:
                f.write(json.dumps({"label": scene_label}))
            data.append((scene_name, scene_label))

data = sorted(data, key=lambda x: int(x[0].replace("scene", "")))
for scene_name, scene_label in data:
    print(template.format(scene_name=scene_name, label=scene_label))

```
