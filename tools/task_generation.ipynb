```python
# create info.json and task config
import os
import json
scene_folder = os.path.join(os.path.expanduser("~"), "data", "isaac_scenes_v1")

template = \
"""      {scene_name}:
        label: {label}
        ceiling_height: 1.0
        episode_number: 20
        navmesh_exclude:
        - /World/ground/terrain/__default_setting/HDR_Sphere/SM_Sphere
        navmesh_preset: default
        path: grscenes_home/scenes/{scene_name}/start_result_navigation.usd
        rule_pattern: gr"""

data_folder = os.path.join(scene_folder, "grscenes_home", "scenes")

count = 0
for scene_name in os.listdir(data_folder):
    if os.path.isdir(os.path.join(data_folder, scene_name)):
        if not scene_name.startswith("scene"):
            scene_label = "none"
            new_scene_name = f'scene{count}'
            os.rename(os.path.join(data_folder, scene_name), os.path.join(data_folder, new_scene_name))
            count += 1

data = []
for scene_name in os.listdir(data_folder):
    if os.path.isdir(os.path.join(data_folder, scene_name)):
        if os.path.exists(os.path.join(data_folder, scene_name, "info.json")):
            info = json.load(open(os.path.join(data_folder, scene_name, "info.json")))
            scene_label = info["label"]
            if scene_label[-1].isdigit():
                scene_label = scene_label[:-1]
            with open(os.path.join(data_folder, scene_name, "info.json"), "w") as f:
                f.write(json.dumps({"label": scene_label}))
        data.append((scene_name, scene_label))

data = sorted(data, key=lambda x: int(x[0].replace("scene", "")))
# randomly pick 15 scenes
import random
data = random.sample(data, 15)
data = sorted(data, key=lambda x: int(x[0].replace("scene", "")))
for scene_name, scene_label in data:
    print(template.format(scene_name=scene_name, label=scene_label))

```

```python
# get episodes that are not bad
import os
import json
scene_folder = os.path.join(os.path.expanduser("~"), "data", "isaac_scenes_v1")
episode_data_folder = os.path.join(scene_folder, "episode_data")

count = 0
for scene_name in os.listdir(episode_data_folder):
    if scene_name.startswith("grCommercial"):
        for episode_name in os.listdir(os.path.join(episode_data_folder, scene_name)):
            if not os.path.isdir(os.path.join(episode_data_folder, scene_name, episode_name)):
                continue
            if not os.path.exists(os.path.join(episode_data_folder, scene_name, episode_name, "generated_instructions.json")):
                print(f"no generated instructions: {scene_name} {episode_name}")
                continue
            data = json.load(open(os.path.join(episode_data_folder, scene_name, episode_name, "generated_instructions.json")))
            if not data.get("is_bad", True):
                episode_name = episode_name.replace("episode_", "")
                print(f"{scene_name}_{episode_name}")
                count += 1
print(count)
```
