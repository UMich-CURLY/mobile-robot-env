```python
# get episodes that are not bad
import os
import json
scene_folder = os.path.join(os.path.expanduser("~"), "data", "isaac_scenes_v1")
episode_data_folder = os.path.join(scene_folder, "episode_data")

count = 0
for scene_name in os.listdir(episode_data_folder):
    if scene_name.startswith("vc"):
        for episode_name in os.listdir(os.path.join(episode_data_folder, scene_name)):
            if not os.path.isdir(os.path.join(episode_data_folder, scene_name, episode_name)):
                continue
            if not os.path.exists(os.path.join(episode_data_folder, scene_name, episode_name, "generated_instructions.json")):
                print(f"no generated instructions: {scene_name} {episode_name}")
                continue
            data = json.load(open(os.path.join(episode_data_folder, scene_name, episode_name, "generated_instructions.json")))
            if not data.get("is_bad", True):
                if not data.get("improved_instructions"):
                    print(f"no improved instructions: {scene_name} {episode_name}")
                    continue
                episode_name = episode_name.replace("episode_", "")
                print(f"{scene_name}_{episode_name}")
                count += 1
print(count)
```

```python
# sameple episodes
import os
import json
import random

episodes = {
    "gr": [],
    "vc": [],
    "innout": [],
}

episode_num = {
    "gr": 150,
    "vc": 150,
    "innout": 100,
}

episode_folder = "../episodes/"
for file_name in os.listdir(episode_folder):
    if file_name.endswith(".txt"):
        for scene_type in episodes.keys():
            if file_name.endswith(scene_type+".txt"):
                data = open(os.path.join(episode_folder, file_name)).readlines()
                data = [x.strip() for x in data]
                valid_scenes = ["vc_boston", "vc_amsterdam", "vc_baltimore", "vc_barcelona", "vc_paris"]
                new_data = []
                for scene in valid_scenes:
                    new_data.extend([x for x in data if scene in x])
                episodes[scene_type].extend(new_data)

for scene_type in episodes.keys():
    sample_num = min(episode_num[scene_type], len(episodes[scene_type]))
    episodes[scene_type] = random.sample(episodes[scene_type], sample_num)
    episodes[scene_type] = sorted(episodes[scene_type])
    for episode_name in episodes[scene_type]:
        print(episode_name)
```
